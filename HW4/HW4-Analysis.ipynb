**Explain how hidden missing values (such as ? symbols) impact raw medical datasets, and justify your choice of imputation strategies for skewed numerical vitals versus categorical diagnostic fields.**

A "?" symbol could be one-hot encoded as a separate category, but it would not have a place in numeric columns. These symbols should be imputed differently. The official dataset from the UCI repository did not include ? symbols because they all appeared to be numeric columns. Thus, I downloaded the other datasets included with the repository and discovered that a number of them had "?" symbols. Here, I included a step in the pipeline that would handle each "?" symbol by removing white space and replacing each with np.nan. This step in the pipeline would then convert all columns to numeric types.

For categorical columns, blanks were replaced with the most frequent value in the imputer. For numeric columns, blanks were replaced with the median value. The distribution of numeric columns appeared close to normal without much left or right skewing. I chose to regularize these columns with standard scaling instead of log-scaling.

According to pg. 152-153 of Chapter 5, the SVC classifier with a RBF kernal creates irregular decision boundaries based on the values of gamma and C.

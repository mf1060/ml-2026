**What challenges did you encounter when loading and inspecting a raw external dataset from Kaggle compared to pre-packaged datasets?**

With a pre-packaged dataset, one can import this dataset from a library without making an external connection to a file. Here, one challenge was importing a CSV file from Kaggle into Google Colab, because the cybersecurity intrusion detection dataset is not hosted on Kaggle with a URL. **How did you determine which columns required missing value imputation versus other data cleaning methods, and what strategy did you choose?**

I used the Pandas `.info()` method to determine if any values were missing and then `.value_counts()` to determine the imputation strategy for columns that had missing values. Here, it turned out that `encryption_method`, a categorical value, had several missing values when using `.info()`. I then used `.value_counts()` on that column to see whether one value could be imputed over the other. For example, one can use SimpleImputer with different [strategies](https://scikit-learn.org/stable/modules/generated/sklearn.impute.SimpleImputer.html) for categorical variables. These strategies include replacing values with the most frequently used value or a constant value like: "unknown." Although AES encryption was the most common, the number of observations with DES encryption came close. It would not be wise to fill in the missing values with one value over the other. Instead, we can fill in the value with a constant like, "unknown," and not bias the dataset.

**Explain how the Scikit-Learn ColumnTransformer and Pipeline streamline the data preparation process and prevent data leakage across preprocessing steps.**

As chapter 2 of our textbook discusses (Aurélien Géron, Hands-On Machine Learning with SciKit-Learn and PyTorch (2025)), pipelines are helpful for defining a step-by-step workflow for preprocessing data. Otherwise, we would have to define variables for each step of this workflow which may inadvertently cause some data to be lost in the process. Pipelines are also helpful for storing features of the datset such as column names so that they can be used to reconstruct a new dataframe.

As page 15 of the Module 2 overview discusses, the ColumnTransformer can combine different steps for preprocessing data, such as processing categorical data differently from numeric data. In the workflow I used, I used a preprocessing pipeline to (1) fill in missing numeric values with the median value and use a standard scaler, (2) fill in missing categorical values with a constant value and transform each categorical value with one-hot encoding, and (3) use a log-scaler for one column, `session_duration`. Without using a pipeline, I would need to create three seperate datasets to perform this data cleaning and then combine them together into one dataframe.

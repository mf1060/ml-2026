**Compare the clustering behavior and cluster assignments of K-Means against DBSCAN and HDBSCAN on the 13-dimensional Wine feature space. Why did DBSCAN and HDBSCAN classify specific wine samples as noise, while K-Means assigned every sample to a cluster?**

According to the [customer_segmentation.ipynb](https://github.com/bforoura/ML26/blob/main/Module8_Unsupervised/customer_segmentation.ipynb) notebook, not all points are assigned to a group. To assign a point to a group this depended on (1) the value of epsilon, which is the distance in value between points in the same "neighborhood" and (2) minimum samples which would compose a cluster (see pg. 256). Because some points would fall outside of this minimum distance, not all points are categorized. Instead, they are labeled as -1. K-Means only measures distance to a cluster relative to all other points. Thus, every point will be assigned to a cluster even if it is very far from a centroid, as long as it is farthest away from the other centroids.

Here, we took a range of epsilon and minimum sample values and graphed them with several DBSCAN models. Even large values for epsilon and minimum values (eps=0.5 and min_samples = 4) included some points as noise. We can see from the K-means analysis that all points are assigned a value, even if they were far away.

According to the [customer_segmentation.ipynb](https://github.com/bforoura/ML26/blob/main/Module8_Unsupervised/customer_segmentation.ipynb) notebook, DBSCAN works better with dense clusters and not as well with sparse clusters. DBSCAN will tend to exclude some relevant points for arbitrary reasons, because they are not close enough to an epsilon value or within a certain number of minimum samples. HDBSCAN will create clusters and segment data based on density differences.

Here, we do not see very much of a performance difference between HDBSCAN and the different DBSCAN models we computed. HDBSCAN still excluded a number of points that were external to the model.

However, HDBSCAN appeared to capture more selective and meaningful groupings of points. When the HDBSCAN model classified some groups into clusters, they appeared to resemble the centroids of the K-means model. The dendogram to the HDBSCAN model revealed limitations to segmenting this dataset which may explain why not all points were classified.

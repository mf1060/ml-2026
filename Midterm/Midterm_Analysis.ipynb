When using stocastic gradient descent to train parameters on these probability scores, we can arrive at better predictions. The model can interpret confident wrong predictions and assign higher penalty scores to those predictions. With higher penalty scores, weights update faster. We can also measure varying performance along decision boundaries and use an ROC curve to better understand the model's performance.

**Explain how implicit missing values impact medical datasets, and justify your choice of imputation strategy when handling these zeros compared to outright row deletion.**

This dataset contained all numeric features but implicit 0s for some values. These implicit 0s would not accurately reflect the features required for classification and thus they needed to be imputed. I used a histogram to determine (1) which values had 0s that should be imputed and (2) which features needed to be standard or log-scaled. By taking a histogram, I determined that the features that required imputation and standard-scaling were `plasma_glucose`, `blood_pressure`, `skin_thickness`, and `bmi`. The 0 values in these features were over-represented and it would make little sense for these features to have 0 values. I determined that `serum_insulin` required imputation but also log-scaling because it had a long tail.

The features that did not require imputation were features where a 0 value would make sense. For instance, `months_pregnant` was not imputed. The columns, `age` and `diabetes_func` did not need to be imputed because the distribution did not show that these columns had 0s, but they can be log-scaled because they have longer tails.

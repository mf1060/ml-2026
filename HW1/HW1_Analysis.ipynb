**Based on the instructional videos, what are the key steps for managing files between Google Colab and GitHub?**

The ideal way to manage files between Google Colab and Github is to connect Google Colab to one's Github repository. When using Google Colab, one can save their notebooks to Github by committing changes made on the web interface to a particular repository. To summarize the steps involved, one would:

1. Create a repository on Github. When creating the repository, one can add a ReadMe (to explain the repository) and a .gitignore (to ignore hidden files that could be accidentally committed) to the repository.
2. Create a subfolder for each homework assignment or create subfolders when saving a notebook in Google Colab (discussed below)
3. Go to Google Colab on the web interface. Navigate to Settings and connect one's Github account to Google Colab.
4. Create a notebook in Google Colab.
5. Save each Colab file wihin a subdirectory, e.g. HW1/HW1.ipynb, select the repository to save, and save the notebook with a commit message.
